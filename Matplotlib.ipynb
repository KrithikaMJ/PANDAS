**MATPLOTLIB**

* It is a python library used for creating graphs and visualizing data
* It is a module used to create and customize graph in python
* plt.plot used to create a line in graph in python
* show Display the graph  

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.plot(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
profit = [5,8,6,12,15]
plt.plot(months,sales,label = "Sales")
plt.plot(months,profit,label = "Profit")
plt.legend()                               # display the names separately in the top most end of the graph as (Sales & Profit) for differentiate the both
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.plot(
    months,
    sales,
    marker="o",
    color="green",
    linestyle="dotted"
)
plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.plot(
    months,
    sales,
    marker="s",
    color="black",
    linestyle="--"
)
plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.bar(months,sales)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.bar(
    months,
    sales,
    color="#FFFF0D"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.scatter(months,sales)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.scatter(
    months,
    sales,
    color="#BC8F8F"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.hist(months)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("Monthly Sales")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
marks = [45,50,52,555,60,62,65,68,
         70,72,75,7,80,82,85,90]
plt.hist(marks)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
categories = ["Food","Travel","Shopping","Others"]
amount = [4000,2000,3000,1000]
plt.pie(amount,labels=categories)
plt.title("Monthly Expenses")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax  = plt.subplots(2,1)
ax[0].plot([1,2,3],[10,20,30])
ax[1].bar(["A","B","C"],[20,15,30])
plt.show()
```
