**loc=label and iloc=position**

```python
data=[30,40,20,50,10]
s=pd.Series(data)
print(s)
```

```python
data=[30,40,20,50,10]
s=pd.Series(data)
print(s.loc[1])
```

```python
data=[30,40,20,50,10]
s=pd.Series(data,index=["A","B","C","D","E"])
print(s.iloc[0])
print(s.loc["B"])
print(s[2:4])
```

```python
a=[1,2,3,4,5]
i=pd.Series(a,index=["a","b","c","d","e"])
print(i.iloc[1:3])
print(i.loc["a":"d"])
```

```python
import pandas as pd
sales = pd.Series( [1200,850,1500,950,2000],
                   index = ["Monday","Tuesday","Wednesday","Thursday","Friday"]
)
print(sales)
```

```python
import pandas as pd
sales = pd.Series( [1200,850,1500,950,2000],
                   index = ["Monday","Tuesday","Wednesday","Thursday","Friday"]
)
print(sales)
print("Maximum sale =",sales.max())
print("Minimum sale =",sales.min())
print("Sum of sales =",sales.sum())
print("Below 1000\n",sales[sales<1000])
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())
print(ar.nunique())
print(ar.value_counts())                          # count for unique
```
